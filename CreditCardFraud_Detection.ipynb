```python
#Time represents seconds elapsed since the first transaction in the dataset.
df['Time_hours'] = df['Time'] / 3600
```

Data Preparation

```python
# splitting the dataset into X and y
# Input features
X = df.drop(
    ["Class", "Time", "Amount"],
    axis=1
)
y = df['Class']

# displaying the shape
print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# train test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train,y_test = train_test_split(X,y, random_state= 42, test_size= 0.2, stratify =y)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
# feaature scaling
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

We use fit_transform() on training data and only transform() on testing data. This prevents data leakage.

Confusion Matrix
True Negative (TN) -- Genuine transactions correctly predicted as genuine.

True Positive (TP)-- Fraud transactions correctly detected.

False Positive (FP) -- Genuine transactions incorrectly flagged as fraud.

False Negative (FN)-- Fraud transactions incorrectly classified as genuine.

Precision: Of all transactions predicted as fraud, how many were actually fraud?'

Recall: Of all actual fraud transactions, how many did the model detect?

F1-score: Balance between precision and recall.



PR-AUC: Measures performance across different precision-recall thresholds.

**Logistic Regression using Cross Validation**

```python
# applying stratified cross validation
from sklearn.model_selection import StratifiedKFold,cross_val_predict
from sklearn.metrics import precision_recall_curve

# lr model
lr_cv = LogisticRegression(max_iter =1000, random_state =42)

# cross validation
cv = StratifiedKFold(n_splits= 5, shuffle = True, random_state =42)

# y_train_probability
y_train_prob =  cross_val_predict(lr_cv,  X_train_scaled, y_train, cv =cv, method = "predict_proba")[:,1]
```

```python
# calculating precision recall at different thresholds
precision, recall, thresholds = precision_recall_curve(
    y_train,
    y_train_prob
)

f1_scores = (
    2 * precision[:-1] * recall[:-1]
    / (precision[:-1] + recall[:-1] + 1e-10)
)

best_index = np.argmax(f1_scores)

best_threshold = thresholds[best_index]

print("Best Threshold:", best_threshold)
print("Precision:", precision[best_index])
print("Recall:", recall[best_index])
print("F1 Score:", f1_scores[best_index])
```

A lower threshold makes the model more willing to flag transactions as fraud. This can increase fraud detection, but it can also increase false positives.

The tuned threshold:
- Improved recall by about 19.62 percentage points.
- Improved F1-score by about 8.18 percentage points.
- Reduced precision by about 7.91 percentage points.

Probability ≥ 0.5 → Fraud (1)



Probability < 0.5 → Genuine (0)

```python
# Training Logistic Regression on complete dataset

# Train on complete training data
lr_cv.fit(X_train, y_train)

# Predict probabilities on test data
test_prob = lr_cv.predict_proba(X_test)[:, 1]

# Apply the threshold selected through cross-validation
best_threshold = 0.07905586565118145
test_pred = (test_prob >= best_threshold).astype(int)

# Evaluate
print(confusion_matrix(y_test, test_pred))
print(classification_report(y_test, test_pred))

print("Test PR-AUC:", average_precision_score(y_test, test_prob))
```

Negative SHAP value: pushes the prediction away from fraud.
